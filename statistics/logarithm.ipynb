```python
data={"income":[2000,3000,4000,5000]}
df=pd.DataFrame(data)
```

```python
df["log_income"]=np.log(df["income"])
print(df)
```

```python
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.hist(df["income"], bins=5)
plt.title("Original Income")
plt.show()
```

```python
plt.subplot(1,2,2)
plt.hist(df["log_income"], bins=5)
plt.title("Log Transformed Income")
plt.show()
```

```python
df["safe_log"] = np.log(df["income"] + 1)
```

```python
df=pd.DataFrame({
    "sales":[10,50,30,70]})
df["log_sales"] = np.log1p(df["sales"])
```

```python
df
```

```python
plt.figure(figsize=(10,4))

# Original
plt.subplot(1,2,1)
plt.hist(df["sales"], bins=10)
plt.title("Original Sales")

# Log transformed
plt.subplot(1,2,2)
plt.hist(df["log_sales"], bins=10)
plt.title("Log Sales")

plt.show()
```
