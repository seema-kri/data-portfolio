```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("bhp.csv")
data.head()
```

```python
data.price_per_sqft.describe()
```

```python
plt.hist(data.price_per_sqft,bins=20,rwidth=0.8)
plt.show()
```

```python
lower_bound=data['price_per_sqft'].quantile(0.001)
lower_bound
```

```python
upper_bound=data['price_per_sqft'].quantile(0.999)
upper_bound
```

```python
outliers=data[(data.price_per_sqft>upper_bound)|(data.price_per_sqft<lower_bound)]
outliers
```

```python
outliers.sample(10)
```

```python
clean_data=data[(data.price_per_sqft<upper_bound)&(data.price_per_sqft>lower_bound)]
clean_data
```
