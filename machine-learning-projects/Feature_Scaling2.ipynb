```python
import pandas as pd
```

```python
df=pd.read_csv('retail_customer.csv')
df.head()
```

```python
df.shape
```

```python
df.dtypes
```

```python
# binning ages into categories
bins=[10,30,50,70]
labels=['Young','Mid','Senior']
df['age_group']=pd.cut(df['age'],bins=bins,labels=labels)
df[['age','age_group']].head(10)
```

```python
# converting dates into datetime
df['signup_date']=pd.to_datetime(df['signup_date'])
df['last_purchase_date']=pd.to_datetime(df['last_purchase_date'])
```

```python
df.dtypes
```

```python
# Time based Feature
df['signup_year']=df['signup_date'].dt.year
df['signup_month']=df['signup_date'].dt.month
df['days_since_signup']=(pd.Timestamp.today()-df['last_purchase_date']).dt.days
```

```python
# Name Features
df['First_Name']=df['name'].apply(lambda x:x.split()[0])
df['Last_Name']=df['name'].apply(lambda x:x.split()[-1])
df['name_length']=df['name'].apply(len)
```

```python
# High Spender Falg
df['is_high_spender']=(df['total_spent']>5000).astype(int)
df.head()
```
