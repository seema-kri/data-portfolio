# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
df_scaled=df.copy()
```

```python
# standard scaler
standard_scaler=StandardScaler()
df_scaled['Salary_StandardScaler']=standard_scaler.fit_transform(df_scaled[['Salary']])
```

```python
# Min-Max Scaler
minmax_scaler=MinMaxScaler()
df_scaled['Salary_MinMaxScaled']=minmax_scaler.fit_transform(df_scaled[['Salary']])
```

```python
df_scaled[['Salary','Salary_StandardScaler','Salary_MinMaxScaled']].head()
```
